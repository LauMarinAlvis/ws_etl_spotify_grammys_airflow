```python
import pandera.pandas as pa
from pandera import Check

cols_0_1 = ["danceability", "energy", "speechiness", "acousticness",
            "instrumentalness", "liveness", "valence"]

spotify_schema = pa.DataFrameSchema(
    {
        "track_id": pa.Column(str, nullable=False),
        "artists": pa.Column(str, nullable=False),
        "album_name": pa.Column(str, nullable=False),
        "track_name": pa.Column(str, nullable=False),
        "popularity": pa.Column(int, Check.in_range(0, 100)),
        "duration_ms": pa.Column(int, Check.gt(0)),
        "explicit": pa.Column(bool),
        "key": pa.Column(int, Check.in_range(-1, 11)),
        "loudness": pa.Column(float, Check.in_range(-60, 5)),
        "mode": pa.Column(int, Check.isin([0, 1])),
        "tempo": pa.Column(float, Check.gt(0)),
        "time_signature": pa.Column(int, Check.in_range(0, 7)),
        "track_genre": pa.Column(str, nullable=False),
        **{c: pa.Column(float, Check.in_range(0, 1)) for c in cols_0_1},
    },
    strict=False,
)

try:
    spotify_schema.validate(spotify, lazy=True)
    print("Validación exitosa Oki Doki")
except pa.errors.SchemaErrors as exc:
    fc = exc.failure_cases
    display(fc.groupby(["column", "check"]).size().reset_index(name="filas_fallidas"))
    filas_malas = fc["index"].dropna().nunique()
    print(f"Filas con algún fallo: {filas_malas} de {len(spotify)} ({filas_malas/len(spotify):.2%})")
```

```python
import unicodedata, re

def normalize_name(s):
    if pd.isna(s):
        return None
    s = unicodedata.normalize("NFKD", str(s)).encode("ascii", "ignore").decode()
    s = re.sub(r"\s+", " ", s).strip().lower()
    return s or None

def paren_names(w):
    if pd.isna(w):
        return []
    return [n for n in (normalize_name(x) for x in re.findall(r"\(([^)]*)\)", w)) if n]

gr = grammys.copy()
gr["artist_norm"] = gr["artist"].map(normalize_name)
gr["paren"] = gr["workers"].map(paren_names)

sp_set = set(spotify["artists"].dropna().str.split(";").explode().map(normalize_name).dropna())

solo_artist = set(gr["artist_norm"].dropna())
con_workers = solo_artist | {n for lst in gr["paren"] for n in lst}

print("Coincidencias solo con columna artist :", len(sp_set & solo_artist))
print("Coincidencias artist + workers        :", len(sp_set & con_workers))
```

```python
#Spotfy x artista
sp = (spotify.dropna(subset=["artists"])
      .drop_duplicates(subset="track_id")                 # canciones que se repiten x genero
      .assign(artist_norm=lambda d: d["artists"].str.split(";"))
      .explode("artist_norm"))
sp["artist_norm"] = sp["artist_norm"].map(normalize_name)

sp_art = (sp.groupby("artist_norm")
            .agg(n_tracks=("track_id", "nunique"),
                 avg_popularity=("popularity", "mean"),
                 avg_danceability=("danceability", "mean"),
                 avg_energy=("energy", "mean"),
                 avg_valence=("valence", "mean"))
            .reset_index())

# grammys x artista columna artist + nombres entre paréntesis de workers
gr["key"] = gr.apply(lambda r: ([r["artist_norm"]] if r["artist_norm"] else []) + r["paren"], axis=1)
gk = (gr.explode("key").dropna(subset=["key"])
        .reset_index().drop_duplicates(["index", "key"]))
gr_art = (gk.groupby("key")
            .agg(grammy_wins=("category", "count"),
                 first_win=("year", "min"),
                 last_win=("year", "max"))
            .reset_index().rename(columns={"key": "artist_norm"}))

# merge
merged = sp_art.merge(gr_art, on="artist_norm", how="left", validate="one_to_one")
merged["grammy_wins"] = merged["grammy_wins"].fillna(0).astype(int)
merged["has_grammy"] = merged["grammy_wins"] > 0

print("Artistas Spotify :", len(sp_art))
print("Artistas Grammys :", len(gr_art))
print("Merge            :", merged.shape, "| con Grammy:", merged["has_grammy"].sum())
display(merged.groupby("has_grammy")[["avg_popularity", "avg_danceability", "avg_energy", "avg_valence"]].mean())
display(merged.sort_values("grammy_wins", ascending=False).head(10))
```

```python
print("Duplicados por (track_id, track_genre):",
      spotify.duplicated(["track_id", "track_genre"]).sum())

EXCLUDE = {"various artists"}
sp_art2 = sp_art[~sp_art["artist_norm"].isin(EXCLUDE)]
gr_art2 = gr_art[~gr_art["artist_norm"].isin(EXCLUDE)]
merged2 = sp_art2.merge(gr_art2, on="artist_norm", how="left", validate="one_to_one")
merged2["grammy_wins"] = merged2["grammy_wins"].fillna(0).astype(int)
merged2["has_grammy"] = merged2["grammy_wins"] > 0
print("Merge final:", merged2.shape, "| con Grammy:", merged2["has_grammy"].sum())
```
